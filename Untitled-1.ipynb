```python
try:
    # Paso 1: Minimizar todas las ventanas activas (equivalente a presionar Win+M)
    pyau.hotkey('win', 'm')

    # Paso 2: Abrir SQL Server Management Studio
    sub.Popen(SQL20)
    print("Ejecutando SQL Server Management Studio...")

    # Paso 3: Esperar dinámicamente a que aparezca la ventana de conexión
    timeout = 120          # Máximo tiempo de espera: 120 segundos
    t0 = time.time()       # Momento en que empezamos a contar el tiempo
    ventana_encontrada = False  # Bandera para saber si encontramos la ventana

    # Bucle que se ejecuta hasta que se acabe el tiempo máximo
    while time.time() - t0 < timeout:
        # Obtener todas las ventanas que tengan este título EXACTO
        ventanas = gw.getWindowsWithTitle("Connect to Server")
        
        # Si la lista de ventanas no está vacía, significa que la encontramos
        if ventanas:
            ventana_encontrada = True
            print("Ventana de conexión detectada.")
            break   # Salir del bucle porque ya encontramos la ventana

        # Si aún no aparece, esperar 1 segundo antes de volver a verificar
        time.sleep(1)

    # Paso 4: Si después del tiempo máximo no se detectó la ventana, lanzar un error
    if not ventana_encontrada:
        raise Exception("No se detectó la ventana de conexión dentro del tiempo esperado.")

    # Paso 5: Hacer una pequeña pausa extra antes de enviar la tecla Enter
    time.sleep(1)

    # Paso 6: Simular presionar Enter para conectar automáticamente
    pyau.press('enter')
    print("Se conectó automáticamente.")

# Capturar cualquier error que ocurra durante el proceso
except Exception as ex:
    print("No se pudo abrir SQL por el error:", ex)
```
